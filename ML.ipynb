```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score, confusion_matrix,
    classification_report, precision_score,
    recall_score, f1_score
)

# 1. Load datasets
train = pd.read_csv('/content/ML_train.csv')
valid = pd.read_csv('/content/ML_valid.csv')
test = pd.read_csv('/content/ML_test.csv')

```

```python
# 2. Separate features and target
drop_cols = ['image_id', 'image_path', 'target']

X_train = train.drop(columns=drop_cols)
X_valid = valid.drop(columns=drop_cols)
X_test = test.drop(columns=drop_cols)

# 3. Encode target labels
y_train = train['target'].map({'Healthy': 0, 'Bleached': 1})
y_valid = valid['target'].map({'Healthy': 0, 'Bleached': 1})
y_test = test['target'].map({'Healthy': 0, 'Bleached': 1})

# 4. Handle missing values
X_train = X_train.replace([np.inf, -np.inf], np.nan)
X_valid = X_valid.replace([np.inf, -np.inf], np.nan)
X_test = X_test.replace([np.inf, -np.inf], np.nan)

medians = X_train.median()

X_train = X_train.fillna(medians)
X_valid = X_valid.fillna(medians)
X_test = X_test.fillna(medians)

# 5. Standardize features
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_valid = scaler.transform(X_valid)
X_test = scaler.transform(X_test)

# 6. Train Logistic Regression
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

# 7. Make predictions
train_pred = model.predict(X_train)
valid_pred = model.predict(X_valid)
test_pred = model.predict(X_test)

# 8. Evaluate the test set
print("Test Accuracy:", accuracy_score(y_test, test_pred))
print("Precision:", precision_score(y_test, test_pred, zero_division=0))
print("Recall:", recall_score(y_test, test_pred, zero_division=0))
print("F1-Score:", f1_score(y_test, test_pred, zero_division=0))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, test_pred))

print("\nClassification Report:")
print(classification_report(
    y_test, test_pred,
    labels=[0, 1],
    target_names=['Healthy', 'Bleached'],
    zero_division=0
))

# 9. Compare dataset performance
print("\nTraining Accuracy:", accuracy_score(y_train, train_pred))
print("Validation Accuracy:", accuracy_score(y_valid, valid_pred))
print("Test Accuracy:", accuracy_score(y_test, test_pred))
```
